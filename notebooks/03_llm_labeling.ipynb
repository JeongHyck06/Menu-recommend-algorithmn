## 4. 균형 잡힌 실험 샘플 추출

약 100개 단위를 뽑는다. 선정 기준은 다음과 같다.

1. 경계 메뉴에서 10개. 반찬 키워드가 있지만 식사로 남은 메뉴라 판단이 애매하다.
2. 나머지 90개를 식품대분류별로 배분한다. 단위 수의 제곱근에 비례해 큰 분류가 전부를 차지하지 않게 한다.
3. 분류 안에서는 비프랜차이즈와 프랜차이즈를 번갈아 뽑는다.
4. 모든 단계에서 대표식품명당 최대 3개. 피자, 버거, 닭튀김이 편중되지 않는다.
5. 상한 때문에 배분량을 못 채우면 남은 풀에서 보충한다.

`random_state=42`로 고정한다. 이미 저장된 `sample_units.csv`가 있으면 그 샘플을 그대로 쓰고 다시 추출하지 않는다. 라벨링 결과와 검토 내용이 샘플에 묶여 있기 때문이다.

## 6. 출력 검증

응답은 아래 JSON이어야 한다.

- 필수 필드 `식품코드`, `라벨`, `근거`
- `식품코드`가 입력과 같아야 한다
- 모든 속성이 있고 값이 허용값 안에 있어야 한다
- 원본 온도(HOT/ICED)와 충돌하는 `제공온도`는 원본을 덮어쓰지 않고 `원본온도충돌` 검토 플래그를 붙인다
- 결과 집합의 누락·중복·대상 외 항목을 확인한다

아래는 검증기 동작을 보여주는 테스트용 예시 문자열이다. 모델 응답이 아니며 결과 파일에 저장하지 않는다.

```python
# 테스트용 예시 (모델 응답 아님)
example_ok = json.dumps({
    "식품코드": example_input["식품코드"],
    "라벨": {"매운맛": "없음", "국물": "국물없음", "제공온도": "뜨거움", "조리법": "구이", "기름짐": "보통", "든든함": "보통"},
    "근거": "생선을 구운 요리",
}, ensure_ascii=False)
validated = validation.validate_response(validation.parse_response_text(example_ok), example_input["식품코드"], example_input["온도_원본"])
print(validated)
```

```python
# 테스트용 실패 예시: 식품코드 불일치, 허용값 밖
for bad in [example_ok.replace(example_input["식품코드"], "D0000"), example_ok.replace('"없음"', '"아주매움"')]:
    try:
        validation.validate_response(validation.parse_response_text(bad), example_input["식품코드"], None)
    except validation.ResponseValidationError as exc:
        print("검증 실패:", exc)
```

## 7. 비용 추정

공식 가격(https://platform.claude.com/docs/en/about-claude/pricing, 2026-09-17 확인): Claude Haiku 4.5 입력 $1 / MTok, 출력 $5 / MTok. Batch API는 50% 할인.

토큰 수는 API 없이 정확히 셀 수 없어 다음 가정을 쓴다.

- 입력 토큰 ≈ (시스템 프롬프트 글자 수 + 사용자 메시지 글자 수) / 1.5. 한국어는 글자당 토큰이 많아 보수적으로 잡았다.
- 출력 토큰 ≈ 150. JSON 라벨 6개와 한 문장 근거.
- 재시도와 검증 실패 재호출은 포함하지 않았다.

키가 준비되면 `client.messages.count_tokens`로 입력 토큰을 정확히 셀 수 있다.

```python
CHARS_PER_TOKEN = 1.5
OUTPUT_TOKENS_PER_UNIT = 150

pricing = runner.PRICING[config.model]
user_chars = sample.apply(lambda r: len(build_user_message(build_unit_input(r))), axis=1)
input_tokens_per_unit = int((len(SYSTEM_PROMPT) + user_chars.mean()) / CHARS_PER_TOKEN)

def cost_table(n_units):
    inp, out = input_tokens_per_unit * n_units, OUTPUT_TOKENS_PER_UNIT * n_units
    return {
        "단위 수": n_units, "입력 토큰(추정)": inp, "출력 토큰(추정)": out,
        "Messages API (USD)": runner.estimate_cost_usd(inp, out, config.model),
        "Batch API (USD)": runner.estimate_cost_usd(inp, out, config.model, batch=True),
    }

print(f"가격 확인일: {pricing['checked_on']}, 입력 ${pricing['input_per_mtok']}/MTok, 출력 ${pricing['output_per_mtok']}/MTok, Batch 할인 {pricing['batch_discount']:.0%}")
print(f"단위당 입력 토큰 추정: {input_tokens_per_unit} (시스템 {len(SYSTEM_PROMPT)}자 + 사용자 평균 {user_chars.mean():.0f}자)")
cost_df = pd.DataFrame([cost_table(len(sample)), cost_table(len(units))], index=["실험 샘플", "전체 식사 단위"])
sample_estimate = cost_df.loc["실험 샘플", "Messages API (USD)"]
print(f"샘플 예상 비용 ${sample_estimate:.4f} / 이번 실행 한도 ${BUDGET_USD:.2f} -> {'한도 안' if sample_estimate <= BUDGET_USD else '한도 초과'}")
cost_df
```

## 8. 샘플 라벨링 실행 (기본 비활성)

`RUN_LIVE_LABELING = True`로 바꾸고 `ANTHROPIC_API_KEY`를 설정한 뒤 실행하면 샘플 단위를 순서대로 호출한다.

- 실행마다 `run_id`를 부여해 이번 실행 비용과 누적 비용을 구분한다.
- 성공 결과는 `results.jsonl`에 바로 추가되어 중단 후 재실행하면 이어서 진행한다.
- 입력 내용, 모델 ID, 프롬프트·스키마 버전이 같은 성공 결과는 다시 호출하지 않고 이번 실행 비용에 넣지 않는다.
- 형식 검증 실패 후 재시도한 응답의 토큰과 비용도 이번 실행에 포함한다.
- 일시적 오류(연결, 429, 5xx)는 최대 `max_retries`회 재시도한다. SDK 내부 재시도는 꺼서(`max_retries=0`) 자체 재시도와 겹치지 않는다.
- 인증·권한·결제 오류는 기록 후 즉시 중단한다.
- 지출 + 다음 호출 최대 예상 비용이 `BUDGET_USD`를 넘기 전에 남은 단위를 `budget_stopped`로 두고 멈춘다.

```python
RESULTS_PATH = LABELING_DIR / "results.jsonl"
store = runner.ResultStore(RESULTS_PATH)
run_id = None

if RUN_LIVE_LABELING:
    run_id = runner.new_run_id()
    print(f"run_id: {run_id}, 모델: {config.model}, 한도: ${BUDGET_USD:.2f}")
    try:
        run_summary = runner.run_labeling(
            sample, config, store, run_id=run_id, budget_usd=BUDGET_USD, input_token_guess=input_tokens_per_unit,
        )
    except runner.FatalLabelingError as exc:
        print("치명적 오류로 중단:", exc)
        run_summary = pd.DataFrame()
    run_config = {**config.__dict__, "run_id": run_id, "budget_usd": BUDGET_USD,
                  "sample_path": SAMPLE_PATH.name, "sample_size": len(sample), "sample_random_state": SAMPLE_RANDOM_STATE}
    (LABELING_DIR / "run_config.json").write_text(json.dumps(run_config, ensure_ascii=False, indent=2), encoding="utf-8")
    if not run_summary.empty:
        print(run_summary["status"].value_counts().to_string())
        print(f"재사용 {int(run_summary['reused'].sum())}건")
    run_stats = runner.summarize_run(store, run_id)
    print(json.dumps(run_stats, ensure_ascii=False, indent=2))
else:
    print("RUN_LIVE_LABELING=False: API를 호출하지 않았다.")
```

## 9. 결과 검증

실제 응답(`mode=live`)만 집계한다. 테스트용 모의 응답(`mode=mock`)은 테스트 코드 안에서만 쓰이고 이 파일에는 들어오지 않는다.

- 단위별 최종 상태와 누락·중복·대상 외 여부
- 실제 토큰 사용량과 추정 비용 (누적)
- JSON 형식 오류, 원본 온도 충돌, 속성별 미확인 비율

```python
records = store.load()
live_records = [r for r in records if r.get("mode") == runner.MODE_LIVE]
mock_records = [r for r in records if r.get("mode") != runner.MODE_LIVE]
print(f"저장된 기록: 실제 호출 {len(live_records)}건, 모의 {len(mock_records)}건")

latest = review.latest_records(live_records)
if latest:
    print("단위별 최종 상태:", pd.Series([r["status"] for r in latest.values()]).value_counts().to_dict())
    print("누락·중복 확인:", {k: len(v) for k, v in validation.check_result_set(list(latest), sample["라벨링단위ID"].tolist()).items()})
    calls = [r for r in live_records if r.get("usage")]
    print(f"누적 호출 {len(calls)}회, 입력 토큰 {sum(r['usage']['input_tokens'] for r in calls):,}, "
          f"출력 토큰 {sum(r['usage']['output_tokens'] for r in calls):,}, 추정 비용 ${sum(r.get('cost_usd') or 0 for r in calls):.4f}")
    print("형식 검증 실패 시도:", sum(r["status"] == runner.STATUS_VALIDATION_FAILED for r in live_records))
else:
    print("실제 응답 없음")
```

```python
successes = [r for r in latest.values() if r["status"] == runner.STATUS_SUCCESS]
if successes:
    label_df = pd.DataFrame([r["labels"] for r in successes])
    print(f"성공 {len(successes)}건 / 원본 온도 충돌 {sum(validation.FLAG_TEMPERATURE_CONFLICT in (r.get('review_flags') or []) for r in successes)}건")
    display(pd.DataFrame({
        "미확인 비율(%)": (label_df == schema.UNKNOWN).mean().mul(100).round(1),
        "최빈값": label_df.mode().iloc[0],
    }))
    for attr in schema.ATTRIBUTES:
        print(f"[{attr}]", label_df[attr].value_counts().to_dict())
else:
    print("성공 결과 없음")
```

## 10. 수동 검토 시트

형식 검증을 통과해도 사실성이 보장되지 않으므로 모든 성공 결과는 `검토대기`로 둔다. 기존 `review_sheet.csv`가 있으면 검토값, 검토상태, 메모를 보존하고 저장 전에 `backups/`에 백업한다.

- 결과가 바뀌지 않은 행(같은 `cache_key`): 이전 검토상태 유지
- 입력·모델·프롬프트·스키마가 바뀐 행: 검토값은 남기되 검토상태를 `재검토`로 표시
- 검토상태 값: `미실행` / `검토대기` / `승인` / `수정` / `재검토` / `실패`

검토자는 `검토_*` 컬럼에 수정값을 적고 `검토상태`를 `승인` 또는 `수정`으로 바꾼다.

```python
REVIEW_PATH = LABELING_DIR / "review_sheet.csv"
existing_review = review.load_review_sheet(REVIEW_PATH)
review_sheet = review.merge_existing_review(review.build_review_sheet(sample, records), existing_review)
backup_path = review.save_review_sheet(review_sheet, REVIEW_PATH)

print("백업:", backup_path.name if backup_path else "없음 (새 파일)")
print("검토상태 분포"); print(review_sheet["검토상태"].value_counts().to_string())
```

### 대표 결과와 검토가 필요한 사례

검토 우선순위: 검토 플래그가 있는 행, 미확인이 2개 이상인 행, 경계 메뉴, 재검토 행.

```python
model_cols = [f"모델_{a}" for a in schema.ATTRIBUTES]
show_cols = ["식품명", "대표식품명", "식품대분류명", "업체명", *model_cols, "근거"]
done = review_sheet[review_sheet["상태"] == runner.STATUS_SUCCESS]
if done.empty:
    print("실제 결과가 없어 표시할 사례가 없다")
else:
    print("대표 결과 (대분류별 1건)")
    display(done.groupby("식품대분류명").head(1)[show_cols])
```

```python
if not done.empty:
    unknown_count = (done[model_cols] == schema.UNKNOWN).sum(axis=1)
    ambiguous_ids = set(sample.loc[sample["선정사유"] == sampling.REASON_AMBIGUOUS, "라벨링단위ID"])
    needs_review = done[
        done["검토플래그"].fillna("").ne("")
        | (unknown_count >= 2)
        | done["라벨링단위ID"].isin(ambiguous_ids)
        | (done["검토상태"] == review.REVIEW_RECHECK)
    ]
    print(f"우선 검토 대상: {len(needs_review)}건")
    display(needs_review[["검토플래그", "검토상태", *show_cols]])
```

## 정리

### 이번 단계에서 한 것

- 라벨 스키마 6개 속성(매운맛, 국물, 제공온도, 조리법, 기름짐, 든든함)과 `미확인` 규칙을 정했다. 재료·알레르기는 스키마에서 제외했다.
- 라벨링 단위를 메뉴명, 접두어, 대표식품명, 대분류, 업체, 온도가 모두 같은 묶음으로 정의했다. 라벨은 단위 밖으로 공유하지 않는다.
- 실험 샘플 100개 단위를 경계 메뉴 10개 + 대분류 배분 + 대표식품명당 3개 상한으로 뽑았다.
- 프롬프트, 요청 파라미터, Batch 요청 구성, 응답 검증, 재시도·캐시·점진 저장, 비용 기록, 검토 시트, 품질 지표를 코드로 준비하고 테스트했다.

### 아직 하지 않은 것

- 실제 API 호출을 하지 않았다. 모델의 라벨 품질은 검증되지 않았다.
- 품질 지표는 모두 `미실행`이다. 실제 실행 후 수동 검토를 거쳐야 값이 나온다.
- 토큰 수는 글자 수 기반 추정이다.

### 실제 샘플 라벨링 실행 방법

1. 터미널에서 `export ANTHROPIC_API_KEY=...`로 키를 설정한다. 노트북에 키를 쓰지 않는다.
2. 기본 설정 셀의 `RUN_LIVE_LABELING = True`로 바꾸거나 환경변수 `RUN_LIVE_LABELING=1`로 실행한다.
3. 노트북을 처음부터 실행한다. 8번 셀이 기존 샘플을 호출하고 `results.jsonl`에 건별로 저장한다. 한도(`BUDGET_USD`) 전에 멈추면 다시 실행해 이어서 진행한다.
4. `review_sheet.csv`의 `검토_*` 컬럼과 `검토상태`를 채운 뒤 노트북을 다시 실행하면 검토 내용이 보존된 채 일치율이 계산된다. 이전 파일은 `backups/`에 남는다.
